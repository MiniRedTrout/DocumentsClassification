```python
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from gensim import corpora, models
from gensim.utils import simple_preprocess
from datasets import load_dataset
import pandas as pd

dataset = load_dataset('emotion')
df_train = pd.DataFrame(dataset['train'])
df_test = pd.DataFrame(dataset['test'])
validation_df = pd.DataFrame(dataset['validation'])

all_texts = list(df_train['text']) + list(df_test['text']) + list(validation_df['text'])

tokenized_documents1= [simple_preprocess(text) for text in all_texts]

dictionary1 = corpora.Dictionary(tokenized_documents1)

bow_corpus1 = [dictionary1.doc2bow(doc) for doc in tokenized_documents1]

lda_model1 = models.LdaModel(bow_corpus1, num_topics=20, id2word=dictionary1, passes=15)

topics = lda_model1.print_topics(num_words=10)
for topic in topics:
    print(topic)

document_topic_vectors = []
for doc_bow in bow_corpus1:
    document_topics = lda_model1.get_document_topics(doc_bow, minimum_probability=0.0)
    document_topic_vector = [topic_prob for _, topic_prob in document_topics]
    document_topic_vectors.append(document_topic_vector)

num_documents_to_print = 5
document_topic_vectors = []
for i, doc_bow in enumerate(bow_corpus1):
    document_topics = lda_model1.get_document_topics(doc_bow, minimum_probability=0.0)
    document_topic_vector = [topic_prob for _, topic_prob in document_topics]
    document_topic_vectors.append(document_topic_vector)
    if i < num_documents_to_print:
        print(f"Document {i + 1} Topic Vector: {document_topic_vector}")
```

```python
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from gensim import corpora, models
from gensim.utils import simple_preprocess

categories = [
    'comp.sys.ibm.pc.hardware',
    'comp.sys.mac.hardware',
    'comp.graphics',
    'comp.windows.x'
]

newsgroups = fetch_20newsgroups(
    subset='all',
    categories=categories,
    shuffle=True,
    random_state=42,
    remove=('headers', 'footers', 'quotes')
)

tokenized_documents2 = [simple_preprocess(text) for text in newsgroups.data]

dictionary2 = corpora.Dictionary(tokenized_documents2)

bow_corpus2 = [dictionary2.doc2bow(doc) for doc in tokenized_documents2]

lda_model2 = models.LdaModel(bow_corpus2, num_topics=20, id2word=dictionary2, passes=15)

topics = lda_model2.print_topics(num_words=10)
for topic in topics:
    print(topic)

document_topic_vectors = []
for doc_bow in bow_corpus2:
    document_topics = lda_model2.get_document_topics(doc_bow, minimum_probability=0.0)
    document_topic_vector = [topic_prob for _, topic_prob in document_topics]
    document_topic_vectors.append(document_topic_vector)

num_documents_to_print = 5
document_topic_vectors = []
for i, doc_bow in enumerate(bow_corpus2):
    document_topics = lda_model2.get_document_topics(doc_bow, minimum_probability=0.0)
    document_topic_vector = [topic_prob for _, topic_prob in document_topics]
    document_topic_vectors.append(document_topic_vector)
    if i < num_documents_to_print:
        print(f"Document {i + 1} Topic Vector: {document_topic_vector}")
```
